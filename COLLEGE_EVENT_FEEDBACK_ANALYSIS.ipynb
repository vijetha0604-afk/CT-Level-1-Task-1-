```python
import pandas as pd
df=pd.read_csv('/content/student_feedback.csv')
df.head()
```

```python
df=df.drop_duplicates()
df=df.dropna(how='all')
df.head()
```

```python
df.describe()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
for col in df.columns[1:]:
  plt.figure(figsize=(6,4))
  sns.countplot(x=col,data=df)
  plt.title(col)
  plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.heatmap(df.corr(), annot=True, cmap='Blues')
plt.title("Correlation Between Rating Questions")
plt.show()
```

```python
df.mean().sort_values(ascending=False)
```

```python
plt.savefig("rating chart.png")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10,5))
sns.barplot(x=df.columns[1:], y=df.mean()[1:].values)
plt.xticks(rotation=90)
plt.title("Average Rating for Each Question")
plt.tight_layout()
plt.savefig("average_ratings.png")
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.countplot(x=df["Degree of difficulty of assignments"])
plt.title("Distribution: Degree of Difficulty of Assignments")
plt.savefig("difficulty_distribution.png")
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.heatmap(df.corr(), annot=True, cmap='Blues')
plt.title("Correlation Between Rating Questions")
plt.savefig("correlation_heatmap.png")
plt.show()
```

```python
df.to_csv("cleaned_feedback.csv", index=False)
```

Key Insights

1. Highest Rated: “Course recommendation based on relevance”
2. Second Highest: “Solves student doubts willingly”
3. Lowest Rated: “Degree of difficulty of assignments” → Needs major improvement.
4. Moderate Ratings: “Use of presentations” and “Structuring of the course”
5. Strong Correlation: Students who rate “Explains concepts well” also rate “Solves doubts willingly” high.



Suggestions for Improvement:-

1. Reduce assignment difficulty or provide more examples step-by-step.
2. Improve course structure by giving a weekly plan or module outline.
3. Increase use of presentations to make classes more interactive.
4. Include more real-life examples while explaining concepts.
5. Add short doubt-solving sessions at end of class.
