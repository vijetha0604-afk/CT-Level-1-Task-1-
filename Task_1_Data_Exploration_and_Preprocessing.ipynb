```python
#Importing Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from google.colab import files

#Upload dataset
uploaded=files.upload()

#Load Dataset
df=pd.read_csv("/Restaturent.csv")
df.head()
```

```python
#Basic dataset info
print("Shape of dataset(rows,columns):",df.shape)
print("\nColumns names:",df.columns.tolist())
print("\nData types:\n",df.dtypes)

#First few rows
print("\nFirst 5 rows:")
print(df.head())

#Missing values
print("\nMissing values in each column:")
print(df.isnull().sum())

#Descriptive statistics
print("\nSummary Statistics(numeric columns only):")
print(df.describe())

cat_cols=df.select_dtypes(include='object').columns
print("\nValue counts for categorical columns:")
for col in cat_cols:
  print(f"n{col} value counts:")
  print(df[col].value_counts())

#Distribution Of Aggregate Rating
plt.figure(figsize=(8,6))
sns.countplot(x='Aggregate rating',data=df,palette="viridis")
plt.title("Distribution of Aggregate Rating")
plt.xlabel("Aggregate Rating")
plt.ylabel("Count")
plt.show()

#Handle Missing Values(fill numeric columns with median)
df_filled=df.fillna(df.median(numeric_only=True))

#Save Cleaned dataset
df_filled.to_csv("Restaturent_Cleaned.csv",index=False)
print("\nCleaned dataset saved as Restaturent_Cleaned.csv")


```
